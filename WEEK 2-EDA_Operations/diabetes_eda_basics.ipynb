```python
import pandas as pd
import numpy as np

                      
df = pd.read_csv("diabetes_prediction_dataset.csv")

                                                                                                          
                                       
                                     

print(df.head())
print(df.tail())
print(df.shape)
print(df.columns)
print(df.dtypes)
df.info()
print(df.describe())

print(df["age"].mean())
print(df["age"].median())
print(df["age"].mode())
print(df["age"].var())
print(df["age"].std())
print(df["age"].min())
print(df["age"].max())
print(df["age"].count())

print(df["smoking_history"].unique())
print(df["smoking_history"].nunique())
print(df["smoking_history"].value_counts())

print(df.isnull())
print(df.isnull().sum())
print(df.isnull().mean() * 100)
df_clean = df.dropna()
df_filled = df.fillna(0)

print(df.duplicated())
print(df.duplicated().sum())
df_no_dups = df.drop_duplicates()

print(df["age"])
print(df[["age", "bmi"]])
print(df.iloc[0:5])
print(df[df["age"] > 30])

print(df.sort_values("age"))
df = df.rename(columns={"age": "Age_Years"})

df["BMI_Adjustment"] = df["bmi"] * 0.10
df = df.drop("BMI_Adjustment", axis=1)

print(df.groupby("gender"))
print(df.groupby("gender")["bmi"].mean())

                                  
df1 = df.iloc[:5]
df2 = df.iloc[5:10]
combined = pd.concat([df1, df2])
print(combined)

                                                                  
df_merge = df.copy()
df_merge["ID"] = np.arange(len(df_merge))
df_left = df_merge.iloc[:5]
df_right = df_merge.iloc[:5][["ID", "bmi", "diabetes"]]
merged = pd.merge(df_left, df_right, on="ID")
print(merged)
```
