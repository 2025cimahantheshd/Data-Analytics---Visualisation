```python
import pandas as pd

data=pd.read_csv(r"C:\Users\user\Downloads\synthetic_dataset.csv")

data.isnull()
```

```python
data.isnull().sum()
```

Mean Imputation

```python
data["Rating"]=data["Rating"].fillna(data["Rating"].mean())
```

Meadian Imputation

```python
data["Price"]=data["Price"].fillna(data["Price"].median())
data.isnull()
```

Forward Fill

```python
data["Stock"]=data["Stock"].bfill()
data.isnull()
```

Encoding Categorical Variable

```python
from sklearn.preprocessing import LabelEncoder

data["Category"]=LabelEncoder().fit_transform(data["Category"])
data.iloc[900:1000]
```

```python
encoded=pd.get_dummies(data,columns=["Category"])
encoded.iloc[300:600]
```

```python
encoded=pd.get_dummies(data,columns=["Category"],dtype=int)
encoded.iloc[200:500]
```
