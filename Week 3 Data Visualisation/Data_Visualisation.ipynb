```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("IRIS.csv")

# Seaborn plots
sns.histplot(df["petal_length"])
sns.histplot(df["sepal_length"])
plt.title("Petal Length and Sepal Length Histogram")
plt.show()

sns.boxplot(data=df[["sepal_length", "sepal_width", "petal_length", "petal_width"]])
plt.title("Box Plot of All Iris Features")
plt.show()

sns.scatterplot(data=df, x="sepal_length", y="petal_length", hue="species")
plt.title("Sepal Length and Petal Length by Species")
plt.show()

sns.pairplot(df, hue="species")
plt.show()

correlation = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].corr()
sns.heatmap(correlation, annot=True)
plt.title("Correlation of Iris Features")
plt.show()

# Matplotlib plots
plt.subplot(1, 2, 1)
plt.hist(df["petal_length"], bins=10)
plt.title("Petal Length")

plt.subplot(1, 2, 2)
plt.hist(df["sepal_length"], bins=10)
plt.title("Sepal Length")
plt.show()

plt.boxplot([df["sepal_length"], df["sepal_width"], df["petal_length"], df["petal_width"]], tick_labels=["Sepal Length", "Sepal Width", "Petal Length", "Petal Width"])
plt.title("Box Plot of All Iris Features")
plt.show()

plt.scatter(df["sepal_length"], df["petal_length"])
plt.title("Sepal Length and Petal Length")
plt.show()

pd.plotting.scatter_matrix(df[["sepal_length", "sepal_width", "petal_length", "petal_width"]])
plt.show()

plt.matshow(correlation)
plt.xticks(range(4), correlation.columns)
plt.yticks(range(4), correlation.columns)
plt.title("Correlation of Iris Features")
plt.show()
```
