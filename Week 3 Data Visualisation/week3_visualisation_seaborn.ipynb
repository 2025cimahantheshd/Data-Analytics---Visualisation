```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("IRIS.csv")

plt.subplot(1, 2, 1)
sns.histplot(df["petal_length"])
plt.title("Petal Length")

plt.subplot(1, 2, 2)
sns.histplot(df["sepal_length"])
plt.title("Sepal Length")
plt.show()

sns.boxplot(data=df[["sepal_length", "sepal_width", "petal_length", "petal_width"]])
plt.title("Box Plot of All Iris Features")
plt.show()

sns.scatterplot(data=df, x="sepal_length", y="petal_length", hue="species")
plt.title("Sepal Length and Petal Length by Species")
plt.show()

sns.pairplot(df, hue="species")
plt.show()

correlation = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].corr()
sns.heatmap(correlation, annot=True)
plt.title("Correlation of Iris Features")
plt.show()
```
